```python
# Without historical emotion
import pandas as pd

# Assuming your dataset is stored in a DataFrame called df
# Replace 'utterance' and 'speaker' with the actual column names in your dataset
# Sort the DataFrame by 'Dialogue_ID' and 'Utterance_ID' to ensure the order is correct
df.sort_values(by=['Dialogue_ID', 'Utterance_ID'], inplace=True)

# Initialize an empty dictionary to store conversation for each dialogue
dialogue_conversations = {}

# Iterate over each row in the DataFrame
for index, row in df.iterrows():
    # Get the current dialogue ID
    dialogue_id = row['Dialogue_ID']

    # If dialogue_id is not in the dialogue_conversations dictionary, initialize an empty list for it
    if dialogue_id not in dialogue_conversations:
        dialogue_conversations[dialogue_id] = []

    # Get the current utterance and speaker
    current_utterance = row['Utterance']
    current_speaker = row['Speaker']

    # Append the current utterance to the conversation list for the current dialogue
    dialogue_conversations[dialogue_id].append(f"{current_speaker}: {current_utterance}")

    # Join the conversation list to create the historical conversation for the current dialogue
    historical_conversation = '\n'.join(dialogue_conversations[dialogue_id])

    # Update the 'Conversation' column with the historical conversation for the current dialogue
    df.at[index, 'Conversation'] = historical_conversation

# Display the DataFrame with the added conversation column
df.head()
```

```python
# with historical emotions
import pandas as pd

# Assuming your dataset is stored in a DataFrame called df
# Replace 'utterance' and 'speaker' with the actual column names in your dataset
# Sort the DataFrame by 'Dialogue_ID' and 'Utterance_ID' to ensure the order is correct
df.sort_values(by=['Dialogue_ID', 'Utterance_ID'], inplace=True)

# Initialize an empty dictionary to store conversation for each dialogue
dialogue_conversations = {}
emotion=''
i=0
# Iterate over each row in the DataFrame
for index, row in df.iterrows():
    # Get the current dialogue ID
    dialogue_id = row['Dialogue_ID']

    # If dialogue_id is not in the dialogue_conversations dictionary, initialize an empty list for it
    if dialogue_id not in dialogue_conversations:
        dialogue_conversations[dialogue_id] = []
        historical_conversation=''
        emotion=''

    # Get the current utterance and speaker
    current_utterance = row['Utterance']
    current_speaker = row['Speaker']

    # Append the current utterance to the conversation list for the current dialogue
    # dialogue_conversations[dialogue_id].append(f"{current_speaker}: {current_utterance}")
    # Join the conversation list to create the historical conversation for the current dialogue
    if emotion!='':
      historical_conversation = f"{historical_conversation}({emotion})\n{current_speaker}: {current_utterance}"
    else:
      historical_conversation=f"{current_speaker}: {current_utterance}"
    # Update the 'Conversation' column with the historical conversation for the current dialogue
    df.at[index, 'Conversation'] = historical_conversation
    emotion=row['Emotion']
# Display the DataFrame with the added conversation column
df.head(10)
```

```python
columns_to_drop = ['Speaker','Sr No.','Utterance_ID','Season','Episode','StartTime','EndTime']
dataset = df.drop(columns=columns_to_drop)
dataset.head()
```

```python
dataset['Instruction'] = """ From this emotions:"joy (1)", "sadness (2)", "fear (3)", "disgust (4)", "surprise (5)" or "neutral (6)" or "anger (7)", what's the predicted emotion of the last speaker in this conversation:"""
```

```python
dataset['Instruction'] = """You should give the emotion that supposed to be expressed by the last speaker of the conversation based on the utterences and the previous emotions of the speakers(human and Aima).\nThe emotion should be one of these emotions : "joy (1)", "sadness (2)", "fear (3)", "disgust (4)", "surprise (5)" or "neutral (6)" or "anger (7)".\nGive the emotion of the last speaker of this conversation without any explainations:"""
```

```python
def generate_prompt(data_point):
    """Gen. input text based on a prompt, task instruction, (context info.), and answer

    :param data_point: dict: Data point
    :return: dict: tokenzed prompt
    """
    # Samples with additional context into.
    if data_point['Conversation']:
        text = f"""<s>[INST]{data_point["Instruction"]}:{data_point["Conversation"]} The emotion of the last speaker is :
 The emotion of the last speaker is :[/INST]</s>{data_point["Emotion"]}</s>"""
    # Without
    else:
        text = f"""<s>[INST]{data_point["Instruction"]} The emotion of the last speaker is :
 The emotion of the last speaker is :[/INST]</s>{data_point["Emotion"]}</s>"""
    return text

# add the "prompt" column in the dataset
text_column = [generate_prompt(data_point) for index, data_point in dataset.iterrows()]
dataset["Text"] = text_column
dataset["prompt"]=text_column
```
