```python
from furhat_remote_api import FurhatRemoteAPI
import time

furhat = FurhatRemoteAPI('localhost')
```

```python
class Gestures:
    Smile = "Smile"
    ExpressSad = "ExpressSad"
    ExpressFear = "ExpressFear"
    ExpressAnger = "ExpressAnger"
    BigSmile = "BigSmile"
    Surprise = "Surprise"
    ExpressDisgust = "ExpressDisgust"
furhat.gesture(Gestures.Smile) # Perform a smile
```

```python
emotion='joy'
result='Hello User, Nice to meet you'
```

```python
#Furhat gestures:
furhat_emotions = {'sadness' : "ExpressSad",
                   'fear' : "ExpressFear",
                   'anger' : "ExpressAnger",
                   'joy' : "BigSmile",
                   'surprise' : "Surprise",
                   'disgust' : "ExpressDisgust"}
```

```python
class Gestures:
    Smile = "Smile"
    ExpressSad = "ExpressSad"
    ExpressFear = "ExpressFear"
    ExpressAnger = "ExpressAnger"
    BigSmile = "BigSmile"
    Surprise = "Surprise"
    ExpressDisgust = "ExpressDisgust"
```

```python

emotion = furhat_emotions.get(emotion, None)

if emotion:
    furhat.gesture(name=emotion)
    furhat.say(text=result)
    
else:
    furhat.say(result)
```
